```python
from IPython.display import display, HTML

# New helper function to get the raw HTML content for a single list
def _get_list_html_content(arr, blue_index=None, red_index=None):
    formatted_items = []

    for i, item in enumerate(arr):
        # Format the item to a 5-character width with right alignment
        item_str = f"{item:>5}"

        # Initialize the style for the inner content (the number itself)
        inner_content_style = ''

        if i == blue_index:
            inner_content_style = 'color: white; background-color: darkblue;'
        elif i == red_index:
            inner_content_style = 'color: white; background-color: red;'

        # Create the inner HTML content, applying background only to the number
        inner_html = f"<span style='{inner_content_style}'>{item_str.strip()}</span>"

        # The outer span ensures a fixed width of 5 characters and right alignment.
        formatted_items.append(f"<span style='display: inline-block; width: 5ch; text-align: right;'>{inner_html}</span>")
    return "[" + ", ".join(formatted_items) + "]"

def print_list_with_formatting(arr, blue_index=None, red_index=None, end_para='\n'):
    # Use the new helper to get the content
    html_content = _get_list_html_content(arr, blue_index, red_index)
    # Append the end_para
    html_output = html_content + end_para
    # Display within a <pre> tag
    display(HTML(f"<pre style='white-space: pre;'>{html_output}</pre>"))
```

```python
from IPython.display import display, HTML

def print_lists_with_formatting(list_of_arr, end_para='\n'):
    # Collect all individual list HTML contents
    all_lists_raw_html_contents = []
    for arr in list_of_arr:
        # Use the helper to get content without any <pre> or end_para
        all_lists_raw_html_contents.append(_get_list_html_content(arr))

    # Join them and wrap the whole thing in a single <pre> tag
    # Add display: inline-block and vertical-align: top to the <pre> for side-by-side display
    final_combined_html_content = "".join([
        f"<pre style='white-space: pre; display: inline-block; vertical-align: top;'>{content}</pre>"
        for content in all_lists_raw_html_contents
    ]) + end_para
    display(HTML(final_combined_html_content))
```

## Merge Sort Algorithmus mit Visualisierung

Dieser Abschnitt implementiert den Merge Sort Algorithmus und visualisiert dessen Schritte (Teilung und Zusammenführung) mithilfe der `print_lists_with_formatting` Funktion.

```python
def merge(left, right):
    """
    Führt zwei sortierte Listen zu einer einzigen sortierten Liste zusammen.
    """
    result = []
    i = j = 0

    # Füge Elemente der Reihe nach hinzu, bis eine der Listen leer ist
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    # Füge verbleibende Elemente hinzu (falls vorhanden)
    result.extend(left[i:])
    result.extend(right[j:])
    return result
```

### `merge` Funktion Erklärung:

Die `merge` Funktion ist ein Hilfsalgorithmus, der zwei bereits sortierte Listen (`left` und `right`) nimmt und sie zu einer einzigen, neuen sortierten Liste zusammenführt. So funktioniert sie:

1.  **Initialisierung**: Es werden ein leeres `result`-Array und zwei Zeiger (`i` und `j`), die beide bei 0 beginnen, erstellt. `i` verfolgt die Position in der `left`-Liste und `j` in der `right`-Liste.
2.  **Vergleich und Zusammenführung**: Solange beide Zeiger innerhalb ihrer jeweiligen Listen sind, werden die Elemente an den aktuellen Zeigerpositionen verglichen:
    *   Ist das Element in `left[i]` kleiner als `right[j]`, wird `left[i]` zum `result`-Array hinzugefügt und `i` wird erhöht.
    *   Andernfalls wird `right[j]` hinzugefügt und `j` wird erhöht.
3.  **Restliche Elemente**: Nachdem eine der Listen vollständig durchlaufen wurde, werden alle verbleibenden Elemente der anderen Liste einfach an das `result`-Array angehängt (da diese Elemente bereits sortiert sind).
4.  **Rückgabe**: Die Funktion gibt die vollständig zusammengeführte und sortierte `result`-Liste zurück.

```python
def merge_sort_visual(arr, level=0):
    """
    Implementiert den Merge Sort Algorithmus mit visueller Ausgabe der Teilungs- und Zusammenführungsschritte.
    """
    if len(arr) <= 1:
        return arr

    # Teilungsphase
    mid = len(arr) // 2
    left_half = arr[:mid]
    right_half = arr[mid:]

    print(f"\n{'  ' * level}Level {level} - Teilen: ")
    # Visualisiere die geteilten Hälften auf einer Zeile
    print_lists_with_formatting([left_half, right_half], end_para='')

    sorted_left = merge_sort_visual(left_half, level + 1)
    sorted_right = merge_sort_visual(right_half, level + 1)

    # Zusammenführungsphase
    merged_list = merge(sorted_left, sorted_right)

    print(f"\n{'  ' * level}Level {level} - Zusammenführen: ")
    # Visualisiere die beiden Hälften und das Ergebnis der Zusammenführung auf einer Zeile
    print_lists_with_formatting([sorted_left, sorted_right, merged_list], end_para='')

    return merged_list
```

### `merge_sort_visual` Funktion Erklärung:

Die `merge_sort_visual` Funktion implementiert den Merge Sort Algorithmus und fügt an entscheidenden Stellen Visualisierungen hinzu:

1.  **Basisfall (`if len(arr) <= 1`)**: Wenn die Liste ein Element oder weniger enthält, ist sie bereits sortiert, und die Rekursion endet.
2.  **Teilungsphase (Splitting)**:
    *   Die Liste wird in der Mitte geteilt (`mid = len(arr) // 2`) in `left_half` und `right_half`.
    *   Es wird eine Ausgabe erzeugt, die den aktuellen Rekursions-`Level` und den Schritt "Teilen" anzeigt.
    *   `print_lists_with_formatting([left_half, right_half], end_para='')` wird aufgerufen, um die beiden Hälften nebeneinander auf einer Zeile zu visualisieren. Der leere `end_para` sorgt dafür, dass sie zusammen bleiben.
    *   Die Funktion ruft sich dann selbst rekursiv für `left_half` und `right_half` auf (`merge_sort_visual(left_half, level + 1)` und `merge_sort_visual(right_half, level + 1)`), wobei der `level` für tiefere Rekursionen erhöht wird.
3.  **Zusammenführungsphase (Merging)**:
    *   Nachdem die rekursiven Aufrufe zurückgekehrt sind (d.h., die `left_half` und `right_half` sind jetzt `sorted_left` und `sorted_right`), werden diese beiden sortierten Listen mithilfe der `merge`-Funktion zusammengeführt.
    *   Es wird eine Ausgabe erzeugt, die den aktuellen `Level` und den Schritt "Zusammenführen" anzeigt.
    *   `print_lists_with_formatting([sorted_left, sorted_right, merged_list], end_para='')` visualisiert die beiden sortierten Eingabelisten und die daraus resultierende `merged_list` nebeneinander.
4.  **Rückgabe**: Die vollständig zusammengeführte und sortierte `merged_list` wird zurückgegeben.

### Level-by-Level Merge Sort Visualisierung

Um eine Visualisierung zu erhalten, die die Aufteilungen und Zusammenführungen Schicht für Schicht (level-by-level oder breadth-first) anzeigt, müssen wir den Prozess leicht anpassen. Statt die Schritte sofort zu drucken, werden sie zuerst gesammelt und dann in der gewünschten Reihenfolge ausgegeben.

```python
def merge_sort_collect_events(arr, level=0):
    """
    Führt den Merge Sort Algorithmus aus und sammelt alle Teilungs- und Zusammenführungsschritte.
    Gibt die sortierte Liste und eine Liste von Ereignissen zurück.
    Jedes Ereignis ist ein Tupel: (level, op_type, *data).
    """
    events = []

    if len(arr) <= 1:
        return arr, events

    mid = len(arr) // 2
    left_half = arr[:mid]
    right_half = arr[mid:]

    # Ereignis für die Teilungsphase hinzufügen
    events.append((level, 'split', list(arr), list(left_half), list(right_half)))

    # Rekursive Aufrufe zum Sortieren der Hälften und Sammeln ihrer Ereignisse
    sorted_left, left_events = merge_sort_collect_events(left_half, level + 1)
    events.extend(left_events)

    sorted_right, right_events = merge_sort_collect_events(right_half, level + 1)
    events.extend(right_events)

    # Zusammenführungsphase
    merged_list = merge(sorted_left, sorted_right)

    # Ereignis für die Zusammenführungsphase hinzufügen
    events.append((level, 'merge', list(sorted_left), list(sorted_right), list(merged_list)))

    return merged_list, events

def visualize_merge_sort_level_by_level(arr):
    """
    Visualisiert den Merge Sort Algorithmus level-by-level.
    """
    print(f"Originale unsortierte Liste: {arr}")

    final_sorted_list, all_events = merge_sort_collect_events(list(arr)) # Wichtig: eine Kopie übergeben

    # Ereignisse nach Level gruppieren
    events_by_level = {}
    for event in all_events:
        level, op_type, *data = event
        if level not in events_by_level:
            events_by_level[level] = []
        events_by_level[level].append((op_type, data))

    max_level = max(events_by_level.keys()) if events_by_level else -1

    print("\n--- Teilungsphase (Splitting) Level für Level ---")
    for level in range(max_level + 1):
        if level in events_by_level:
            split_events_at_level = [e for e in events_by_level[level] if e[0] == 'split']
            if split_events_at_level:
                print(f"\n{'  ' * level}Level {level} - Teilen:")
                combined_split_lists_for_level = []
                for op_type, data in split_events_at_level:
                    original_list, left_half, right_half = data
                    combined_split_lists_for_level.extend([left_half, right_half])
                if combined_split_lists_for_level:
                    print_lists_with_formatting(combined_split_lists_for_level, end_para='')

    print("\n--- Zusammenführungsphase (Merging) Level für Level ---")
    for level in range(max_level, -1, -1): # Zusammenführungen von tiefsten Level zum höchsten
        if level in events_by_level:
            merge_events_at_level = [e for e in events_by_level[level] if e[0] == 'merge']
            if merge_events_at_level:
                print(f"\n{'  ' * level}Level {level} - Zusammenführen:")
                combined_merge_lists_for_level = []
                for op_type, data in merge_events_at_level:
                    sorted_left, sorted_right, merged_result = data
                    combined_merge_lists_for_level.extend([sorted_left, sorted_right, merged_result])
                if combined_merge_lists_for_level:
                    print_lists_with_formatting(combined_merge_lists_for_level, end_para='')

    print(f"\nEndgültig sortierte Liste: {final_sorted_list}")
'OK'
```

### Erklärung der neuen Funktionen:

1.  **`merge_sort_collect_events(arr, level=0)`**:
    *   Diese Funktion ist eine Modifikation des ursprünglichen `merge_sort_visual`. Sie führt immer noch den rekursiven Merge Sort aus, aber anstatt die Schritte direkt zu drucken, sammelt sie sie in einer Liste von `events`.
    *   Jedes `event` ist ein Tupel, das den `level` der Operation, den `op_type` ('split' oder 'merge') und die beteiligten Listen enthält. Dies ermöglicht eine spätere flexible Ausgabe.
    *   Sie gibt sowohl die endgültig sortierte Liste als auch die gesammelten Ereignisse zurück.

2.  **`visualize_merge_sort_level_by_level(arr)`**:
    *   Dies ist die neue Hauptfunktion für die Visualisierung.
    *   Sie ruft `merge_sort_collect_events` auf, um alle Ereignisse zu erhalten.
    *   Anschließend gruppiert sie diese Ereignisse nach ihrem `level`.
    *   **Ausgabe der Teilungsphase**: Sie iteriert von `Level 0` bis zum tiefsten Level und druckt für jedes Level alle gesammelten 'split'-Ereignisse. Dies zeigt, wie die Liste Schicht für Schicht aufgeteilt wird.
    *   **Ausgabe der Zusammenführungsphase**: Danach iteriert sie vom tiefsten Level zurück zu `Level 0` und druckt für jedes Level alle gesammelten 'merge'-Ereignisse. Dies visualisiert, wie die sortierten Teillisten wieder zusammengeführt werden.
    *   Die `print_lists_with_formatting` Funktion wird weiterhin verwendet, um die Listen ordentlich nebeneinander darzustellen.

```python
def bubble_sort_kurz(arr, print_output=False, end_p="\n"): # Changed default end_p to newline for general use
    original_arr = list(arr) # Kopie der Original-Liste für die Ausgabe
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                swapped = True
                # Highlight the elements that were just swapped
                print_list_with_formatting(arr, blue_index=j, red_index=j+1, end_para='') # Keep swaps on one line
        if not swapped:
            break

    if print_output:
        display(HTML("<b><br>Originale und Sortierte Liste:</b>"))
        # Use print_lists_with_formatting to display both lists side-by-side
        print_lists_with_formatting([original_arr, arr], end_para=end_p)

    return arr

# Demonstration der Funktion mit print_output=True
my_list_demo = [-5, 3, 2, 1, -3, -3, 7, 2, 2]
print("--- Demonstration der sortierten Liste mit formatierter Ausgabe ---")
sorted_result = bubble_sort_kurz(list(my_list_demo), print_output=True, end_p=' ---- ENDE\n')

print(f"\nDer Rückgabewert der Funktion ist (sortiert): {sorted_result}")
```

## Bubble Sort mit detaillierter Schritt-für-Schritt-Visualisierung

Diese Funktion visualisiert jeden Schritt des Bubble Sort Algorithmus, indem sie die Liste **vor** und **nach** jedem Tauschvorgang nebeneinander anzeigt. Hierfür wird die `print_lists_with_formatting` Funktion verwendet.

```python
def bubble_sort_with_lists_formatting_steps(arr):
    current_arr = list(arr) # Eine Kopie erstellen, um die Original-Liste nicht zu ändern
    n = len(current_arr)

    print(f"Originale Liste: {arr}")
    print("\n--- Schritt-für-Schritt Bubble Sort Visualisierung ---")

    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            # Zustand vor dem potenziellen Tausch erfassen
            arr_before_swap = list(current_arr)

            if current_arr[j] > current_arr[j+1]:
                current_arr[j], current_arr[j+1] = current_arr[j+1], current_arr[j]
                swapped = True
                print(f"\nDurchlauf {i+1}, Vergleich {j+1}: Tausch {arr_before_swap[j]} und {current_arr[j]} - Vorher | Nachher")
                print_lists_with_formatting([arr_before_swap, current_arr], end_para='')
            # Optional: Zeige auch Schritte ohne Tausch
            # else:
            #     print(f"\nDurchlauf {i+1}, Vergleich {j+1}: Keine Tausch - Aktuell")
            #     print_lists_with_formatting([current_arr], end_para='')

        if not swapped:
            break

    print(f"\n--- Sortierung abgeschlossen ---")
    print_lists_with_formatting([arr, current_arr], end_para=' ---- ENDE\n')
    return current_arr
```

### Erklärung der `bubble_sort_with_lists_formatting_steps` Funktion:

Diese Funktion basiert auf dem Standard Bubble Sort Algorithmus, wurde aber um eine erweiterte Visualisierung erweitert:

1.  **`current_arr = list(arr)`**: Erstellt eine Arbeitskopie der übergebenen Liste, damit die Original-Liste unverändert bleibt.
2.  **Äußere Schleife (`for i in range(n)`)**: Steuert die Anzahl der Durchläufe. Nach jedem vollständigen Durchlauf ist das größte Element im unsortierten Teil der Liste an seiner korrekten Endposition.
3.  **Innere Schleife (`for j in range(0, n - i - 1)`)**: Iteriert durch den unsortierten Teil der Liste, vergleicht benachbarte Elemente.
4.  **`arr_before_swap = list(current_arr)`**: Vor jeder potenziellen Tauschoperation wird der aktuelle Zustand der Liste `current_arr` in `arr_before_swap` gespeichert. Dies ist entscheidend für die "Vorher"-Visualisierung.
5.  **Tausch-Bedingung (`if current_arr[j] > current_arr[j+1]`)**: Wenn ein Element größer ist als sein Nachfolger, werden sie getauscht.
6.  **Visualisierung des Tauschs**: Wenn ein Tausch stattfindet:
    *   Es wird eine Textausgabe generiert, die angibt, welcher Durchlauf und Vergleich stattfindet und welche Elemente getauscht werden.
    *   `print_lists_with_formatting([arr_before_swap, current_arr], end_para='')` wird aufgerufen. Diese Funktion stellt die Liste **vor dem Tausch** und **nach dem Tausch** nebeneinander auf einer Zeile dar. Der `end_para=''` verhindert einen Zeilenumbruch direkt nach der Visualisierung, sodass nachfolgende Ausgaben nahtlos anschließen können.
7.  **Optimierung (`if not swapped: break`)**: Wenn in einem Durchlauf keine Tauschvorgänge stattfinden, ist die Liste bereits sortiert, und der Algorithmus wird vorzeitig beendet.
8.  **Endgültige Visualisierung**: Nach Abschluss des Sortierprozesses wird die originale Liste und die endgültig sortierte Liste ebenfalls nebeneinander angezeigt.

```python
print("\n--- Demonstration der Schritt-für-Schritt Bubble Sort Visualisierung ---")
my_list_for_step_visual = [5, 1, 4, 2, 8]
sorted_list_steps = bubble_sort_with_lists_formatting_steps(my_list_for_step_visual)
print(f"\nDer Rückgabewert der Funktion ist (sortiert): {sorted_list_steps}")
```

## Schritt-für-Schritt Insertion Sort Visualisierung

Diese Funktion visualisiert jeden Schritt des Insertion Sort (Beszúrásos rendezés) Algorithmus, indem sie den Zustand der Liste nach der korrekten Platzierung jedes Elements im sortierten Teil anzeigt. Es werden die `print_list_with_formatting` und `print_lists_with_formatting` Funktionen verwendet, um die Änderungen deutlich zu machen.

```python
def insertion_sort_with_steps_visualization(arr):
    n = len(arr)
    current_arr = list(arr)  # Eine Kopie erstellen, um die Original-Liste nicht zu ändern

    print(f"Originale Liste: {arr}")
    print("\n--- Schritt-für-Schritt Insertion Sort Visualisierung ---")

    # Initialer Zustand (erstes Element wird als sortiert betrachtet)
    print("\nInitialer Zustand (erstes Element als sortiert betrachtet):")
    print_list_with_formatting(current_arr)

    for i in range(1, n):
        key = current_arr[i]
        j = i - 1

        print(f"\nSchritt {i} (Index {i}): Element {key} soll in den sortierten Teil eingefügt werden.")
        # Aktuellen Zustand zeigen, mit dem Element, das eingefügt werden soll, hervorgehoben.
        print("Aktueller Zustand (Element zum Einfügen hervorgehoben):")
        print_list_with_formatting(current_arr, blue_index=i)

        # Verschiebe Elemente von current_arr[0..i-1], die größer als 'key' sind,
        # um eine Position nach vorne.
        while j >= 0 and key < current_arr[j]:
            current_arr[j + 1] = current_arr[j]
            j -= 1
        current_arr[j + 1] = key

        # Zustand nach der Einfügung zeigen, mit dem neu eingefügten Element hervorgehoben.
        print(f"Zustand nach Einfügung von {key} an Position {j+1}:")
        print_list_with_formatting(current_arr, red_index=j+1)

    print(f"\n--- Sortierung abgeschlossen ---")
    print_list_with_formatting(current_arr, end_para=' ---- ENDE\n')
    return current_arr
```

## Schritt-für-Schritt Selection Sort Visualisierung

Diese Funktion visualisiert jeden Schritt des Selection Sort Algorithmus. In jedem Durchlauf wird das kleinste Element im unsortierten Teil der Liste gefunden und an die korrekte Position am Anfang des unsortierten Teils verschoben.

Die Visualisierung zeigt:
1.  Den Ausgangszustand der Liste.
2.  Für jeden Durchlauf (`i`):
    *   Die Suche nach dem Minimum im unsortierten Teil.
    *   Den Zustand der Liste, wobei die aktuell zu besetzende Position (Index `i`) und das gefundene Minimum (Index `min_idx`) hervorgehoben werden, **bevor** der Tausch stattfindet.
    *   Den Zustand der Liste **nach** dem Tausch, wobei die neu platzierten Elemente hervorgehoben sind.

Hierfür werden die bereits definierten Formatierungsfunktionen `print_list_with_formatting` und `print_lists_with_formatting` verwendet, um die Änderungen klar darzustellen.

```python
from IPython.display import display, HTML

# Modifizierte print_lists_with_formatting, um separate Highlight-Indizes zu akzeptieren
def print_lists_with_formatting(list_of_arr, list_of_blue_indices=None, list_of_red_indices=None, end_para='\n'):
    # Collect all individual list HTML contents
    all_lists_raw_html_contents = []
    for k, arr in enumerate(list_of_arr):
        blue_idx = list_of_blue_indices[k] if list_of_blue_indices and k < len(list_of_blue_indices) else None
        red_idx = list_of_red_indices[k] if list_of_red_indices and k < len(list_of_red_indices) else None
        # Use the helper to get content without any <pre> or end_para
        all_lists_raw_html_contents.append(_get_list_html_content(arr, blue_index=blue_idx, red_index=red_idx))

    # Join them and wrap the whole thing in a single <pre> tag
    # Add display: inline-block and vertical-align: top to the <pre> for side-by-side display
    final_combined_html_content = "".join([
        f"<pre style='white-space: pre; display: inline-block; vertical-align: top;'>{content}</pre>"
        for content in all_lists_raw_html_contents
    ]) + end_para
    display(HTML(final_combined_html_content))

def selection_sort_with_steps_visualization(arr):
    n = len(arr)
    current_arr = list(arr)  # Eine Kopie erstellen, um die Original-Liste nicht zu ändern

    print(f"Originale Liste: {arr}")
    print("\n--- Schritt-für-Schritt Selection Sort Visualisierung ---")

    print("\nInitialer Zustand:")
    print_list_with_formatting(current_arr)

    for i in range(n - 1):
        min_idx = i
        print(f"\nSchritt {i+1} (Index {i}): Suche nach dem kleinsten Element im unsortierten Teil ({current_arr[i:]}).")
        print("Aktueller unsortierter Teil (Beginn hervorgehoben in Blau):")
        print_list_with_formatting(current_arr, blue_index=i) # Highlight the starting point of unsorted part

        # Finde das kleinste Element im unsortierten Teil
        for j in range(i + 1, n):
            if current_arr[j] < current_arr[min_idx]:
                min_idx = j

        # Wenn das kleinste Element nicht bereits an der richtigen Position ist, tausche es
        if min_idx != i:
            print(f"Kleinste Element ({current_arr[min_idx]}) gefunden bei Index {min_idx}. Tausche mit Element bei Index {i} ({current_arr[i]}):")

            arr_before_swap = list(current_arr)
            current_arr[i], current_arr[min_idx] = current_arr[min_idx], current_arr[i]

            # Visualisiere vor und nach dem Tausch
            # Vorher: Highlighte die Elemente, die getauscht werden
            # Nachher: Highlighte die Positionen, wo die Elemente gelandet sind
            print_lists_with_formatting(
                [arr_before_swap, current_arr],
                list_of_blue_indices=[i, i],      # Index i in beiden Listen blau
                list_of_red_indices=[min_idx, min_idx], # Index min_idx in beiden Listen rot
                end_para=''
            )
        else:
            print(f"Element bei Index {i} ({current_arr[i]}) ist bereits das kleinste im verbleibenden unsortierten Teil.")
            print_list_with_formatting(current_arr, blue_index=i)

    print(f"\n--- Sortierung abgeschlossen ---")
    print_list_with_formatting(current_arr, end_para=' ---- ENDE\n')
    return current_arr
```

```python
def counting_sort_collect_events(arr):
    """
    Führt den Counting Sort Algorithmus aus und sammelt alle Schritte als Ereignisse.
    Gibt die sortierte Liste und eine Liste von Ereignissen zurück.
    Jedes Ereignis ist ein Tupel: (step_type, *data).
    """
    events = []
    n = len(arr)

    if n == 0:
        events.append(('initial_state', []))
        return [], events

    # 1. Maximales Element finden, um die Größe des Zähl-Arrays zu bestimmen
    max_val = 0
    if arr: # Sicherstellen, dass die Liste nicht leer ist, bevor max() aufgerufen wird
        max_val = max(arr)
    events.append(('max_element_found', list(arr), max_val))

    # 2. Zähl-Array (Frequenz-Array) initialisieren
    count = [0] * (max_val + 1)
    events.append(('count_array_initialized', list(arr), list(count)))

    # 3. Zähl-Array füllen: Häufigkeit jedes Elements zählen
    for num in arr:
        count[num] += 1
    events.append(('counts_populated', list(arr), list(count)))

    # 4. Zähl-Array modifizieren: Kumulative Summen bilden (Positionsindizes)
    for i in range(1, max_val + 1):
        count[i] += count[i - 1]
    events.append(('cumulative_counts', list(arr), list(count)))

    # 5. Ausgabe-Array initialisieren
    output = [0] * n
    events.append(('output_array_initialized', list(arr), list(output)))

    # 6. Ausgabe-Array bauen: Elemente von rechts nach links in die sortierten Positionen einfügen
    # Iteration von rechts nach links gewährleistet Stabilität
    for i in range(n - 1, -1, -1):
        num = arr[i]
        output_index = count[num] - 1
        output[output_index] = num
        count[num] -= 1  # Zähler für das platzierte Element dekrementieren

        # Ereignis für das Platzieren eines Elements
        events.append(('place_element', list(arr), list(output), num, i, output_index, list(count)))

    # 7. Finales sortiertes Array
    sorted_arr = list(output)
    events.append(('final_sorted', list(sorted_arr)))

    return sorted_arr, events

def visualize_counting_sort_steps(arr_original):
    """
    Visualisiert den Counting Sort Algorithmus Schritt-für-Schritt.
    """
    print(f"Originale unsortierte Liste: {arr_original}")
    arr = list(arr_original) # Eine Kopie erstellen, um die Original-Liste nicht zu ändern

    sorted_arr, all_events = counting_sort_collect_events(arr)

    print("\n--- Counting Sort Visualisierung (Schritt-für-Schritt) ---")

    for event in all_events:
        event_type = event[0]

        if event_type == 'initial_state':
            current_list = event[1]
            print(f"\nInitialer Zustand der Liste:")
            print_list_with_formatting(current_list)

        elif event_type == 'max_element_found':
            current_list, max_val = event[1], event[2]
            print(f"\nMaximales Element in der Liste gefunden: {max_val}. Dies bestimmt die Größe des Zähl-Arrays.")
            print_list_with_formatting(current_list)

        elif event_type == 'count_array_initialized':
            current_list, count_array = event[1], event[2]
            print(f"\nZähl-Array (Frequenz-Array) initialisiert mit Nullen (Größe {len(count_array)}):")
            print_list_with_formatting(count_array)
            print("Original-Liste (für Kontext):")
            print_list_with_formatting(current_list)

        elif event_type == 'counts_populated':
            current_list, count_array = event[1], event[2]
            print(f"\nZähl-Array nach dem Zählen der Elemente aus der Original-Liste:")
            print_list_with_formatting(count_array)
            print("Original-Liste (für Kontext):")
            print_list_with_formatting(current_list)

        elif event_type == 'cumulative_counts':
            current_list, count_array = event[1], event[2]
            print(f"\nZähl-Array nach dem Berechnen der kumulativen Summen (Positionsindizes):")
            print_list_with_formatting(count_array)
            print("Original-Liste (für Kontext):")
            print_list_with_formatting(current_list)

        elif event_type == 'output_array_initialized':
            current_list, output_array = event[1], event[2]
            print(f"\nAusgabe-Array initialisiert mit Nullen (Größe {len(output_array)}):")
            print_list_with_formatting(output_array)
            print("Original-Liste (für Kontext):")
            print_list_with_formatting(current_list)

        elif event_type == 'place_element':
            original_arr_snapshot, current_output_array, num_to_place, original_idx, output_idx, current_count_array = event[1], event[2], event[3], event[4], event[5], event[6]
            print(f"\nPlatziere Element {num_to_place} (von Original-Index {original_idx}) in das Ausgabe-Array an Index {output_idx}.")
            print("Zustand des Ausgabe-Arrays:")
            print_list_with_formatting(current_output_array, red_index=output_idx) # Hebt das neu platzierte Element hervor
            print("Aktueller Stand des Zähl-Arrays (relevant für das Element):")
            print_list_with_formatting(current_count_array, blue_index=num_to_place) # Hebt den Zähler für die platzierte Zahl hervor
            print("Original-Liste (für Kontext):")
            print_list_with_formatting(original_arr_snapshot, blue_index=original_idx) # Hebt das aktuell genommene Element hervor

        elif event_type == 'final_sorted':
            final_list = event[1]
            print(f"\n--- Sortierung abgeschlossen ---")
            print_list_with_formatting(final_list)

    print(f"\nEndgültig sortierte Liste: {sorted_arr}")
    return sorted_arr
```
