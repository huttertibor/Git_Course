```markdown
# Bubble Sort Algorithmus

Bubble Sort ist ein einfacher Sortieralgorithmus, der wiederholt durch die Liste geht, um benachbarte Elemente zu vergleichen und zu vertauschen, wenn sie in der falschen Reihenfolge sind. Der Durchlauf durch die Liste wird so oft wiederholt, bis keine Vertauschungen mehr vorgenommen werden müssen, was bedeutet, dass die Liste sortiert ist.

Obwohl Bubble Sort einfach zu verstehen und zu implementieren ist, ist er für große Datenmengen ineffizient. Seine Zeitkomplexität ist im besten Fall (Liste ist bereits sortiert) O(n) und im schlechtesten und durchschnittlichen Fall O(n²).
```

````python
def bubble_sort(arr):
    n = len(arr)
    # Durchlaufe alle Elemente der Liste
    for i in range(n):
        # 'swapped' wird verwendet, um zu überprüfen, ob in diesem Durchlauf
        # Vertauschungen stattgefunden haben. Wenn nicht, ist die Liste sortiert.
        swapped = False

        # Letzte 'i' Elemente sind bereits an der richtigen Stelle sortiert.
        # Daher müssen wir nur bis 'n-i-1' vergleichen.
        for j in range(0, n-i-1):

            # Vergleiche das aktuelle Element mit dem nächsten Element
            if arr[j] > arr[j+1]:
                # Wenn das aktuelle Element größer ist, vertausche sie
                arr[j], arr[j+1] = arr[j+1], arr[j]
                swapped = True

        # Wenn in diesem Durchlauf keine zwei Elemente vertauscht wurden,
        # ist die Liste bereits sortiert und wir können abbrechen.
        if not swapped:
            break
    return arr
````

```markdown
## Erklärung der `bubble_sort` Funktion:

1.  **`bubble_sort(arr)`**: Definiert eine Funktion namens `bubble_sort`, die eine Liste `arr` als Eingabe erwartet.
2.  **`n = len(arr)`**: Ermittelt die Länge der Liste und speichert sie in der Variable `n`.
3.  **Äußerer Loop (`for i in range(n)`):** Dieser Loop läuft `n` Mal durch die Liste. Jedes Mal, wenn der äußere Loop abgeschlossen ist, ist das größte unsortierte Element an seiner korrekten Position am Ende des unsortierten Teils der Liste platziert.
4.  **`swapped = False`**: Eine boolesche Variable, die am Anfang jedes äußeren Durchlaufs auf `False` gesetzt wird. Sie wird verwendet, um zu optimieren: Wenn in einem kompletten inneren Durchlauf keine Vertauschung stattfindet, bedeutet dies, dass die Liste bereits sortiert ist und wir den Algorithmus vorzeitig beenden können.
5.  **Innerer Loop (`for j in range(0, n-i-1)`):** Dieser Loop vergleicht benachbarte Elemente und vertauscht sie, falls sie in der falschen Reihenfolge sind. Die `-i-1` im `range` stellt sicher, dass wir nur den unsortierten Teil der Liste betrachten, da die letzten `i` Elemente bereits sortiert sind.
6.  **`if arr[j] > arr[j+1]:`**: Überprüft, ob das aktuelle Element (`arr[j]`) größer ist als das nächste Element (`arr[j+1]`).
7.  **`arr[j], arr[j+1] = arr[j+1], arr[j]`**: Wenn die Bedingung erfüllt ist, werden die beiden Elemente vertauscht. Python ermöglicht diese einfache Zuweisung, um Werte zu tauschen.
8.  **`swapped = True`**: Wenn eine Vertauschung stattfindet, wird `swapped` auf `True` gesetzt.
9.  **`if not swapped: break`**: Nach jedem inneren Durchlauf wird überprüft, ob `swapped` `False` geblieben ist. Wenn ja, bedeutet dies, dass keine Vertauschungen stattgefunden haben, die Liste vollständig sortiert ist und der äußere Loop mit `break` beendet werden kann.
10. **`return arr`**: Die Funktion gibt die sortierte Liste zurück.
```

````python
my_list = [-5, 3, 2, 1, -3, -3, 7, 2, 2]
print(f"Originale Liste: {my_list}")

sorted_list = bubble_sort(my_list)
print(f"Sortierte Liste: {sorted_list}")
````

```markdown
## Demonstration des Bubble Sort Algorithmus

In diesem Abschnitt wird die `bubble_sort` Funktion mit der von Ihnen bereitgestellten Liste `[-5, 3, 2, 1, -3, -3, 7, 2, 2]` getestet.

1.  **Definition der Liste**: Die Variable `my_list` wird mit den vorgegebenen Zahlen initialisiert.
2.  **Ausgabe der Original-Liste**: Die unsortierte Liste wird zur Veranschaulichung ausgegeben.
3.  **Aufruf der Sortierfunktion**: Die `bubble_sort` Funktion wird mit `my_list` aufgerufen, und das Ergebnis (die sortierte Liste) wird in `sorted_list` gespeichert.
4.  **Ausgabe der Sortierten Liste**: Die endgültige, sortierte Liste wird angezeigt.
```

````python
def bubble_sort_kurz(arr, print_output=False):
    original_arr = list(arr) # Kopie der Original-Liste für die Ausgabe
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                swapped = True
        if not swapped:
            break

    if print_output:
        print("Originale Liste (kurz):")
        formatted_original = [f"{item:>5}" for item in original_arr]
        print("[" + ", ".join(formatted_original) + "]")

        print("\nSortierte Liste (kurz) mit gleicher Textbreite:")
        formatted_sorted = [f"{item:>5}" for item in arr]
        print("[" + ", ".join(formatted_sorted) + "]")

    return arr

# Demonstration der aktualisierten Funktion
my_list_to_sort = [-5, 3, 2, 1, -3, -3, 7, 2, 2]

print("\n--- Demonstration ohne formatierte Ausgabe ---")
sorted_result_no_print = bubble_sort_kurz(list(my_list_to_sort))
print(f"Sortierte Liste (Rückgabe ohne direkten Druck): {sorted_result_no_print}")

print("\n--- Demonstration mit formatierter Ausgabe ---")
sorted_result_with_print = bubble_sort_kurz(list(my_list_to_sort), print_output=True)
````
